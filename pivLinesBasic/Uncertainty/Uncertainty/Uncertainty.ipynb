```python
import scipy.io
import matplotlib.pyplot as plt
import numpy as np
import os
```

```python
# Put all the mat files in a list for easy access
df = "./Uncertainty.mat"
mat = scipy.io.loadmat(df)

type(mat)
uncertainty = mat['U'][0][0]['m60']
vxu = uncertainty[0][0]['vx']

xc = mat['U'][0][0]['x']
yc = mat['U'][0][0]['y']
vxu.shape

```

```python
# x-vel (horizontal velocity)
vx = uncertainty[0][0]['vx']
nx = vx.shape[0]
ny = vx.shape[1]

# y-vel
vy = uncertainty[0][0]['vy']
assert vy.shape[0] == nx, "what? size inconsistency"
assert vy.shape[1] == ny, "what? size inconsistency"

# y-vel
vz = uncertainty[0][0]['vz']
assert vz.shape[0] == nx, "what? size inconsistency"
assert vz.shape[1] == ny, "what? size inconsistency"
```

```python
#xmin = -1
#xmax =  1
#xlev = np.linspace(xmin,xmax,128)
#xtic = np.linspace(xmin,xmax,9)

#ymin = -3
#ymax =  3
#ylev = np.linspace(ymin,ymax,128)
#ytic = np.linspace(ymin,ymax,9)

#zmin = -16
#zmax =  16
#zlev = np.linspace(zmin,zmax,128)
#ztic = np.linspace(zmin,zmax,9)

fig, ax = plt.subplots(1,3,figsize=(18,10))

# x-velocity
cnt = ax[0].contourf(xc.reshape(nx), yc.reshape(ny), vx.transpose(),levels=33,cmap='viridis')
for c in cnt.collections:
    c.set_edgecolor("face")

cbar = plt.colorbar(cnt, ax=ax[0])
#cbar.set_ticks(xtic)
#cbar.set_label(r"$v_x$ [m/s]", fontsize=18)

ax[0].set_ylabel('Axial position [mm]', fontsize=18)
ax[0].set_xlabel('Radial position [mm]', fontsize=18)
ax[0].set_title('Radial velocity [m/s]', fontsize=20)

# y-velocity
#cnt = ax[1].contourf(vy.transpose(),levels=np.linspace(0, 1, 33),cmap='viridis')
cnt = ax[1].contourf(xc.reshape(nx), yc.reshape(ny), vy.transpose(),levels=33,cmap='viridis')
for c in cnt.collections:
    c.set_edgecolor("face")

cbar = plt.colorbar(cnt, ax=ax[1])
#cbar.set_ticks(ytic)
##cbar.set_label(r"$v_y$ [m/s]", fontsize=18)
#
ax[1].set_xlabel('Radial position [mm]', fontsize=18)
ax[1].set_title('Axial velocity [m/s]', fontsize=20)
#    
# z-velocity

#cnt = ax[2].contourf(vz.transpose(),levels=np.linspace(0, 3, 33),cmap='viridis')
cnt = ax[2].contourf(xc.reshape(nx), yc.reshape(ny), vz.transpose(),levels=33,cmap='viridis')
for c in cnt.collections:
        c.set_edgecolor("face")

cbar = plt.colorbar(cnt, ax=ax[2])
#cbar.set_ticks(ztic)
##cbar.set_label(r"$v_z$ [m/s]", fontsize=18)
#
ax[2].set_xlabel('Radial position [mm]', fontsize=18)
ax[2].set_title('Azimuthal velocity [m/s]', fontsize=20)
#
#plt.savefig('zerodeg.png',density=300)
plt.show()
```
